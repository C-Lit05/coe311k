```python
## Just went and copy-pasted the whole dang thing, hope that's what you intended
import math
import random
import matplotlib.pyplot as plt
import numpy as np

def montecarlo(sample_num):
    in_circ = 0
    for _ in range(sample_num):
        x = random.uniform(0,1)
        y = random.uniform(0,1)
        if x**2 + y**2 <= 1:
            in_circ += 1
            
    return in_circ * 4 / sample_num
## done with setting up the montecarlo and now setting up pi for the both approximation and the error calculation
realpi = round(math.pi, 15)

abs_ers = []
rel_ers = []

num_run = 100
num_samp = 1000
## I do not understand why these are necessary for the plotting. Is it because it plots 1 by 1 rather than all at once?
## If you have the time I'd appreciate the insight since I can't attend regular office hours with my schedule

for _ in range(100): ## 100 runs
    pi_est = montecarlo(1000) ## 1000 samples
    abs_error = abs(realpi - pi_est)
    rel_error = abs((realpi - pi_est) / realpi)
    
    abs_ers.append(abs_error)
    rel_ers.append(rel_error)
    
plt.figure(figsize=(8,5))
plt.plot(range(1, num_run + 1), abs_ers, marker='o', linestyle='None')
plt.xlabel("Run Number")
plt.ylabel("Absolute Error")
plt.title("Absolute Error of Monte Carlo π")
plt.show()

plt.figure(figsize=(8,5))
plt.plot(range(1, num_run + 1), rel_ers, marker='o', linestyle='None')
plt.xlabel("Run Number")
plt.ylabel("Relative Error")
plt.title("Relative Error of Monte Carlo π")
plt.show()

    
    
```
