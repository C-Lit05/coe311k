```python
import numpy as np


def jacobi(A, b, x0=None, tol=1e-10, max_iter=10000):
    n = len(b)
    x = np.zeros(n) if x0 is None else x0.copy()
    D = np.diag(A)
    R = A - np.diagflat(D)

    for k in range(max_iter):
        x_new = (b - R @ x) / D
        
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1
        
        x = x_new

    return x, max_iter

def gauss(A, b, x0=None, tol=1e-10, max_iter=10000):
    n = len(b)
    x = np.zeros(n) if x0 is None else x0.copy()

    for k in range(max_iter):
        x_old = x.copy()

        for i in range(n):
            sum1 = np.dot(A[i, :i], x[:i])
            sum2 = np.dot(A[i, i+1:], x_old[i+1:])
            x[i] = (b[i] - sum1 - sum2) / A[i, i]

        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            return x, k+1

    return x, max_iter

A1 = np.array([
    [0.582745, 0.48,     0.10,     0.,       0.],
    [0.48,     1.044129, 0.46,     0.10,     0.],
    [0.10,     0.46,     1.10431,  0.44,     0.10],
    [0.,       0.10,     0.44,     0.963889, 0.42],
    [0.,       0.,       0.10,     0.42,     0.522565]
], dtype=float)

b1 = np.array([1.162745, 2.084129, 2.20431, 1.923889, 1.042565], dtype=float)

x_jacobi, it_j = jacobi(A1, b1)
x_gs, it_gs = gauss(A1, b1)

## should give similar or same solution
## Gauss is noticeably quicker

print("For Array Set 1:")
print("Jacobi:", x_jacobi, "Iterations:", it_j)
print("Gauss-Seidel:", x_gs, "Iterations:", it_gs)

A2 = np.array([[1, 1], [1, 1.0001]])
b2 = np.array([1, 1])

x_j2, it_j2 = jacobi(A2, b2)
x_gs2, it_gs2 = gauss(A2, b2)

## Again Gauss is the GOAT

print()
print("For Array Set 2:")
print("Jacobi:", x_j2, "Iterations:", it_j2)
print("Gauss-Seidel:", x_gs2, "Iterations:", it_gs2)

b3 = np.array([1, 1.0001])

x_j3, it_j3 = jacobi(A2, b3)
x_gs3, it_gs3 = gauss(A2, b3)

## Gauss actually struggled with this one

print()
print("For Array Set 3:")
print("Jacobi:", x_j3, "Iterations:", it_j3)
print("Gauss-Seidel:", x_gs3, "Iterations:", it_gs3)
```
