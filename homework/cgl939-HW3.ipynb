```python
import numpy as np
import matplotlib.pyplot as plt
import math as m

def taylorsin(x,termn):
    result = 0
    for n in range(termn):
        ## thank you chatgpt for the .factorial i was not aware of
        term = ((-1)**n) * (x**(2*n + 1)) / m.factorial(2*n + 1)
        result += term
    return result

xvals = np.arange(0,10.1,0.1)

exactvals = np.sin(xvals)
errors = {}

for termn in [1,2,3,4]:
    approxvals = np.array([taylorsin(x, termn) for x in xvals])
    error = np.abs(exactvals - approxvals)
    errors[termn] = error
    
plt.figure(figsize=(10,6))

for termn in [1,2,3,4]:
    plt.plot(xvals, errors[termn], label=f'{termn} term(s)')
    
plt.xlabel('x')
plt.ylabel('Absolute Error')
plt.title('Truncation Error')
plt.legend()
plt.show()

## the more terms added, the more accurate the estimate becomes
## thus the error decreases with each term included. 
## larger values of x have a larger error, while smaller values have less.
## this is because the Taylor series converges faster near 0
```
