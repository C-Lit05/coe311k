```python
import numpy as np
import matplotlib.pyplot as plt

def lstsq(x, y):
    n = len(x)

    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_xy = np.sum(x*y)
    sum_x2 = np.sum(x**2)

    m = (n*sum_xy - sum_x*sum_y) / (n*sum_x2 - sum_x**2)
    b = (sum_y - m*sum_x) / n

    return m, b

np.random.seed(0)

X_rand = np.random.rand(20)
Y_rand = np.random.rand(20)


sorted_indices = np.argsort(X_rand)
X_rand = X_rand[sorted_indices]
Y_rand = Y_rand[sorted_indices]


m_custom, b_custom = lstsq(X_rand, Y_rand)


A = np.vstack([X_rand, np.ones(len(X_rand))]).T
m_numpy, b_numpy = np.linalg.lstsq(A, Y_rand, rcond=None)[0]

print("Random Data Test")
print("Custom method: ", m_custom, b_custom)
print("NumPy method:  ", m_numpy, b_numpy)
```
