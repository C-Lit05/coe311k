```python
import numpy as np
import matplotlib.pyplot as plt


def stability_factor(h, x):
    return np.abs(1 - 2 * h * np.cos(x))

x_vals = np.linspace(0, 10 * np.pi, 1000)
h_to_check = [0.1, 0.5, 0.8, 1.0, 1.1]

plt.figure(figsize=(10, 6))

for h in h_to_check:
    factor = stability_factor(h, x_vals)
    plt.plot(x_vals, factor, label=f'h = {h}')


plt.axhline(y=1.0, color='r', linestyle='--', label='Stability Limit (1.0)')

plt.title("Euler Stability Analysis: $|1 - 2h\cos(x)| \leq 1$")
plt.xlabel("x")
plt.ylabel("Stability Factor")
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()

#Extract the coefficient of y, -2cos(x).
# Determine largest possible value of eigenvalue stuff
# Apply Euler Formula
# For this ODE step size must be less than or equal to 1 to prevent the solution from growing infinitely due to numerical error
```
