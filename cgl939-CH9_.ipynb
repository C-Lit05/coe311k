```python
import numpy as np
import matplotlib.pyplot as plt

# Data points 
X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

# Stack the design matrix
A = np.vstack([X, np.ones(len(X))]).T

# Solve using least squares
m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")

# Plot
plt.scatter(X, Y, color='blue', label='Data Points')

x_line = np.linspace(min(X), max(X), 100)
y_line = m * x_line + c

## approx y = .98x + .14

plt.plot(x_line, y_line, color='red', label='Best Fit Line')
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.title("Least Squares Best Fit Line")
plt.show()

np.random.seed(0)

X_rand = np.random.rand(5) * 10
Y_rand = np.random.rand(5) * 10

A_rand = np.vstack([X_rand, np.ones(len(X_rand))]).T
m_rand, c_rand = np.linalg.lstsq(A_rand, Y_rand, rcond=None)[0]

plt.scatter(X_rand, Y_rand, color='blue', label='Random Data')

x_line = np.linspace(min(X_rand), max(X_rand), 100)
y_line = m_rand * x_line + c_rand

plt.plot(x_line, y_line, color='red', label='Best Fit Line')
plt.legend()
plt.title("Least Squares on Random Data")
plt.show()

## everything fits worse now

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)


X = np.arange(0, 1.01, 0.01)

# True equation: y = 5x - 1
true_m = 5
true_c = -1


noise = np.random.normal(0, 0.3, len(X))
Y = true_m * X + true_c + noise


A = np.vstack([X, np.ones(len(X))]).T
m_est, c_est = np.linalg.lstsq(A, Y, rcond=None)[0]

print("Estimated slope:", m_est)
print("Estimated intercept:", c_est)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X, Y, s=10, label='Noisy Data')
plt.plot(X, true_m * X + true_c, 'g', linewidth=2, label='Original Line (5x - 1)') ## i hate plotting. chatgpt handled the .plot process
plt.plot(X, m_est * X + c_est, 'r', linewidth=2, label='Least Squares Fit')
plt.legend()
plt.title("Least Squares Recovery of y = 5x - 1")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
