```python
import numpy as np
import matplotlib.pyplot as plt

# Step size
a = 0
b = 2 * np.pi
N = 100
x = np.linspace(a, b, N)
h = (b - a) / (N - 1)

def f(x):
    return np.sin(x)

df_true = np.cos(x)
# Central Difference Approximation for the derivative

df_dx = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
### YOUR CODE HERE ###
df_forward = (f(x + h) - f(x)) / h

 

# Backward Difference Approximation
df_backward = (f(x) - f(x - h)) / h
### YOUR CODE HERE

 

# Plot the approximation and the true derivative

fig, axs = plt.subplots(2, 2, figsize=(12, 10))
fig.suptitle('Comparison of Numerical Differentiation Methods', fontsize=16)


axs[0, 0].plot(x, df_true, color='black', label='cos(x)')
axs[0, 0].set_title('True Derivative (Analytical)')
axs[0, 0].legend()


axs[0, 1].plot(x, df_true, 'k--', alpha=0.3, label='True')  ## Gemini AI was instrumental in learning how this type of graphing works
axs[0, 1].plot(x, df_forward, color='red', label='Forward Diff')
axs[0, 1].set_title('Forward Difference')
axs[0, 1].legend()


axs[1, 0].plot(x, df_true, 'k--', alpha=0.3, label='True')
axs[1, 0].plot(x, df_backward, color='blue', label='Backward Diff')
axs[1, 0].set_title('Backward Difference')
axs[1, 0].legend()


axs[1, 1].plot(x, df_true, 'k--', alpha=0.3, label='True')
axs[1, 1].plot(x, df_dx, color='green', label='Central Diff')
axs[1, 1].set_title('Central Difference')
axs[1, 1].legend()


for ax in axs.flat:
    ax.set(xlabel='x', ylabel="f'(x)")
    ax.grid(True)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()

## Do a canvas of 4 subplots or 4 graphs on one plot comparing cos(x), Forward Difference, Backwards Difference, and Central Difference Methods

```
