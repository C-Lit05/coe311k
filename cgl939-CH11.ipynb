```python
import numpy as np

import matplotlib.pyplot as plt

from scipy.interpolate import CubicSpline

 

# Define known data points

x = np.array([0, 1, 2, 3, 4, 5])

y = np.array([0, 1, 0, 1, 0, 1])

# Create a cubic spline interpolation of the data

cs = CubicSpline(x, y, bc_type='natural')# Generate new x values for a smooth plot

x_interp = np.linspace(0, 5, 100)

y_interp = cs(x_interp)


##Use the above code to compare Polifit and CubicSpline  

deg = 5
coeffs = np.polyfit(x, y, deg)
poly = np.poly1d(coeffs)

y_poly = poly(x_interp)
y_spline = cs(x_interp)

plt.plot(x, y, 'o', label='Data Points')
plt.plot(x_interp, y_poly, '--', label='Polynomial Fit')
plt.plot(x_interp, y_spline, label='Cubic Spline')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Polynomial Fit vs Cubic Spline")
plt.legend()
plt.show()

## first off, idk what you want when you say "compare" so
## I just made a graph
## polyfit fits one poly to all data pots while
## CubicSpline fits piecwise polynomials btw each pair
## Spline is goated
## There's a comparison if you wanted it in word form. 

## I am having some weird bug where 
## running the code occasionally gives me just a description
## of the graph and not the actual thing
## not to big of an issue just thought
## I'd let you know
```
