```python
import numpy as np
import matplotlib.pyplot as plt

def solve_oscillator(k, h, x_end=10):
    x_values = np.arange(0, x_end + h, h)
    y = np.zeros(len(x_values))
    v = np.zeros(len(x_values))
    
 
    y[0] = 1.0
    v[0] = 0.0
    
    for i in range(1, len(x_values)):

        
    
        y[i] = y[i-1] + h * v[i-1]
        v[i] = v[i-1] + h * (-k * y[i-1])
        
    return x_values, y


k_val = 1.0
h_list = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

plt.figure(figsize=(12, 6))
for h in h_list:
    x_vals, y_vals = solve_oscillator(k_val, h)
    plt.plot(x_vals, y_vals, label=f'h={h}')

plt.title(f'Harmonic Oscillator Stability Test (k={k_val})')
plt.xlabel('x')
plt.ylabel('Position y(x)')
plt.axhline(0, color='black', lw=1)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()

#Yes, it is inherently unstable. In a perfect harmonic oscillator, the energy should be conserved.
#High k increases the speed of the oscillating. 
#No. For oscillatory systems like this, Forward Euler is a poor choice for large h.
```
