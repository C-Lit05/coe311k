```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score

np.random.seed(0)

x = np.linspace(-5, 5, 50)
y = 0.5*x**3 - 2*x**2 + x + np.random.normal(0, 10, len(x))


A = np.vstack([x, np.ones(len(x))]).T
m, b = np.linalg.lstsq(A, y, rcond=None)[0]
y_linear = m*x + b


poly_coeff = np.polyfit(x, y, 3)
y_poly = np.polyval(poly_coeff, x)


mse_linear = mean_squared_error(y, y_linear)
r2_linear = r2_score(y, y_linear)

mse_poly = mean_squared_error(y, y_poly)
r2_poly = r2_score(y, y_poly)

print("Linear Fit:")
print("MSE:", mse_linear)
print("R²:", r2_linear)

print("\nCubic Polynomial Fit:")
print("MSE:", mse_poly)
print("R²:", r2_poly)


if mse_poly < mse_linear:
    print("\nPolynomial fit is better")
else:
    print("\nLinear fit is better")


plt.scatter(x, y, label="Data")
plt.plot(x, y_linear, label="Linear Fit", color="red")
plt.plot(x, y_poly, label="Cubic Polynomial", color="green")
plt.legend()
plt.title("Least Squares vs Polynomial Fit")
plt.show()

degrees = range(1, 16)

mse_list = []
r2_list = []

for d in degrees:
    
    coeff = np.polyfit(x, y, d)
    y_pred = np.polyval(coeff, x)
    
    mse = mean_squared_error(y, y_pred)
    r2 = r2_score(y, y_pred)
    
    mse_list.append(mse)
    r2_list.append(r2)
    
    print(f"Degree {d}")
    print("MSE:", mse)
    print("R²:", r2)
    print()

    
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(degrees, mse_list, marker='o')
plt.title("MSE vs Polynomial Degree")
plt.xlabel("Degree")
plt.ylabel("MSE")

plt.subplot(1,2,2)
plt.plot(degrees, r2_list, marker='o')
plt.title("R² vs Polynomial Degree")
plt.xlabel("Degree")
plt.ylabel("R²")

plt.tight_layout()
plt.show()

degree = 3


V = np.vander(x, degree + 1)


coeff_vander = np.linalg.lstsq(V, y, rcond=None)[0]


y_vander = V @ coeff_vander


mse_vander = mean_squared_error(y, y_vander)
r2_vander = r2_score(y, y_vander)

print("Vandermonde Polynomial Fit")
print("MSE:", mse_vander)
print("R²:", r2_vander)


plt.scatter(x, y, label="Data")
plt.plot(x, y_vander, label="Vandermonde Polynomial", color="purple")
plt.legend()
plt.title("Polynomial Fit via Vandermonde Matrix")
plt.show()

##I HATE THIS ASSIGNMENT BUT I HEART YOU KIND GRADER
```
