```python
import numpy as np
import time
import matplotlib.pyplot as plt


def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    
    return C


matrix_numbers = list(range(1, 9))
naive_times = []
numpy_times = []
sizes = []

current_size = 3

print(f"{'Matrix #':<10} | {'Size':<10} | {'Naive (s)':<12} | {'NumPy (s)':<12}")
print("-" * 55)

for i in matrix_numbers:
    
    A = np.random.rand(current_size, current_size)
    B = np.random.rand(current_size, current_size)
    
    
    A_list = A.tolist()
    B_list = B.tolist()
    start_naive = time.perf_counter()
    _ = naive_matrix_multiplication(A_list, B_list)
    end_naive = time.perf_counter()
    
    
    start_numpy = time.perf_counter()
    _ = np.dot(A, B)
    end_numpy = time.perf_counter()
    
    
    naive_times.append(end_naive - start_naive)
    numpy_times.append(end_numpy - start_numpy)
    sizes.append(current_size)
    
    print(f"{i:<10} | {current_size:<10} | {naive_times[-1]:<12.6f} | {numpy_times[-1]:<12.6f}")
    
    
    current_size *= 2


plt.figure(figsize=(10, 6))
plt.plot(matrix_numbers, naive_times, marker='o', label='Naive (Lecture Code)')
plt.plot(matrix_numbers, numpy_times, marker='s', label='NumPy (Optimized)')

plt.yscale('log')  
plt.title('In-Class Challenge: Matrix Multiplication Runtimes')
plt.xlabel('Matrix Pair Number (Increasing Size)')
plt.ylabel('Runtime (Seconds, log scale)')
plt.xticks(matrix_numbers)
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.show()

## The custom function is complex because of the 3 nested loops for each dimension
## NumPy algorithms are really good with the large datasets.
## Like in the big O  the O grows much faster than simpler complexitites. The other function is thus impractical. 
```
